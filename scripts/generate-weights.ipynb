```python
import pandas as pd
import numpy as np

# Columns "word" and "count" should NEVER be changed
possible_answers_filename = "../internal/wordle/corpus/possible_answers.csv"

def get_possible_answers(filename):
    return pd.read_csv(filename)

def save_weights(df, filename):
    df.to_csv(filename, index=False)

def add_log_weights(df):
    df["weight"] = np.ceil(np.log10(df["count"]))
    # subtract by min - 1 so that min weight is 1
    min_weight = df["weight"].min() - 1
    df["weight"] = df["weight"] - min_weight
    return df

possible_answers_df = get_possible_answers(possible_answers_filename)
possible_answers_df = add_log_weights(possible_answers_df)
save_weights(possible_answers_df, possible_answers_filename)
```
